```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
health_df = health_df.rename(columns={'City Code': 'City_Code'})
merged_df = pd.merge(weather_df, air_df, on=['Date', 'City_Code'], how='inner')
master_df = pd.merge(merged_df, health_df, on=['Date', 'City_Code'], how='inner')
```

```python
print(master_df.columns.tolist())
```

```python
sns.scatterplot (data=master_df, x='AQI_y', y='Respiratory_Admissions_EstCases', hue='City_Name')

plt.title('Air Quality Index(AQI) vs Respiratory Admissions')
plt.xlabel('AQI Level')
plt.ylabel('Respiratory Admissions')
plt.grid(True)
plt.legend(title = 'Cities')
plt.show()
```

```python
city_aqi = master_df.groupby('City_Name')['AQI_y'].mean().sort_values(ascending=False).reset_index()
```

```python

sns.barplot(data=city_aqi, x='City_Name', y='AQI_y', hue='City_Name', palette='RdYlGn', legend=False)

plt.title('Avg Air Quality Index(AQI) by City')
plt.xlabel('City Name')
plt.ylabel('Average AQI')
plt.xticks(rotation=45)

plt.show
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(data=master_df, x='Respiratory_Admissions_EstCases', fill=True, color='dodgerblue', alpha=0.5, linewidth=2.5)

plt.title('Density Plot of Estimated Respiratory Admissions', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Estimated Respiratory Cases', fontsize=12)
plt.ylabel('Density (Probability)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.kdeplot(data=master_df, x='AQI_y', hue='City_Name', fill=True, common_norm=False, palette='Set1', alpha=0.3, linewidth=2)

plt.title('AQI Density Comparison Across Cities', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('AQI Level', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
